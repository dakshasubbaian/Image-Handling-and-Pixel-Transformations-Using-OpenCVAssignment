14. Read the image ('boy.jpg').

```python
boy = cv2.imread("boy.jpg",1)
boy.shape
boy = boy[:,:,::-1]
plt.imshow(boy)
```

15. Adjust the brightness of the image.

```python
# Create a matrix of ones (with data type float64)
# matrix_ones = 
# YOUR CODE HERE
matrix_one = np.ones(boy.shape, dtype= "uint8")*20
```

16. Create brighter and darker images.

```python
#img_brighter = cv2.add(img, matrix)
#img_darker = cv2.subtract(img, matrix)
# YOUR CODE HERE
img_brighter = cv2.add(boy, matrix_one)
img_darker = cv2.subtract(boy, matrix_one)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
plt.figure(figsize = [25, 10])
plt.subplot(141)
plt.title("Original Image")
plt.imshow(boy)

plt.subplot(142)
plt.title("Darker Image")
plt.imshow(img_darker)

plt.subplot(143)
plt.title("Brighter Image")
plt.imshow(img_brighter)
```

18. Modify the image contrast.

```python
# Create two higher contrast images using the 'scale' option with factors of 1.1 and 1.2 (without overflow fix)
# matrix1 = 
# matrix2 = 
# img_higher1 = 
# img_higher2 = 
# YOUR CODE HERE
matrix1 = np.ones(boy.shape, dtype="float32") * 1.1
matrix2 = np.ones(boy.shape, dtype="float32") * 1.2
img_higher1 = boy / matrix1
img_higher2 = boy * matrix2
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
plt.figure(figsize = [25, 10])
plt.subplot(141)
plt.title("Original Image")
plt.imshow(boy)

plt.subplot(142)
plt.title("Lower Contract Image")
plt.imshow(img_higher1)

plt.subplot(143)
plt.title("Higher Contrast Image")
plt.imshow(img_higher2)
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
boy = cv2.imread("boy.jpg",1)
b, g, r = cv2.split(boy)
zeros = np.zeros_like(b)
```

21. _merged the R, G, B , display along with orginal image 

```python
plt.figure(figsize = [25, 20])
plt.subplot(241)
plt.imshow(cv2.merge([r, zeros, zeros]))
plt.title("Red Channel")

plt.subplot(242)
plt.imshow(cv2.merge((zeros, g, zeros)))
plt.title("Green Channel")

plt.subplot(141)
plt.imshow(cv2.merge((zeros, zeros, b)))
plt.title("Blue Channel")

img_merge = cv2.merge((r, g, b))
plt.subplot(142)
plt.imshow(img_merge)
plt.title("Merged Channel")
```

22. Split the image into the H, S, V components &  Display the channels.

```python
boy_hsv = cv2.cvtColor(boy, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(boy_hsv)
plt.imshow(boy_hsv)
```

23. _merged the H, S, V, display along with orginal image 

```python
plt.figure(figsize = [25, 20])
plt.subplot(241)
plt.imshow(cv2.merge([h, zeros, zeros]))
plt.title("HUE Channel")

plt.subplot(242)
plt.imshow(cv2.merge((zeros, s, zeros)))
plt.title("Saturation Channel")

plt.subplot(141)
plt.imshow(cv2.merge((zeros, zeros, v)))
plt.title("Intensity Channel")

img_merge = cv2.merge((h, s, v))
plt.subplot(142)
plt.imshow(cv2.cvtColor(img_merge, cv2.COLOR_HSV2RGB))
plt.title("Merged Channel")
```
